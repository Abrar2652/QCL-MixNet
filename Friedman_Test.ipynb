```python
# Refined and complete code including Nemenyi test with full print statements

import numpy as np
from scipy.stats import f
import itertools

# Model average ranks (input)
model_names = [
    "QCL-MixNet", "SVM (SMOTE)", "KNN", "RF", "CNN", "GB", "BiGRU", "GRU", "BiLSTM",
    "Balanced RF", "ResNet", "GraphSAGE", "MLP", "XGBoost", "LSTM", "GAT", "GIN", "JKNet",
    "DT", "GCN", "LR"
]
average_ranks = np.array([
    3.55, 4.35, 5.70, 7.18, 8.75, 8.98, 9.52, 10.42, 10.58, 10.62,
    10.62, 10.75, 11.20, 11.45, 13.08, 14.58, 14.72, 14.92, 16.10, 16.50, 17.05
])

# Step 1: Setup
k = len(average_ranks)
N = 4  # number of metrics
alpha = 0.05

# Step 2: Friedman Chi-square using formula
sum_Rj_squared = np.sum(average_ranks ** 2)
chi2_f = ((12 * N) / (k * (k + 1))) * sum_Rj_squared - (3 * N * (k + 1))

# Step 3: FF-statistic
numerator = (N - 1) * chi2_f
denominator = (N * (k - 1)) - chi2_f
F_F = numerator / denominator if denominator != 0 else np.inf
p_value = 1 - f.cdf(F_F, numerator, denominator)

# Step 4: Critical F-value (from table)
df1 = k - 1
df2 = (k - 1) * (N - 1)
F_critical = f.ppf(1 - alpha, df1, df2)
reject_null = F_F > F_critical

# Step 7: Print all results
print("=== Friedman Test (Manual) ===")
print(f"Number of classifiers (k): {k}")
print(f"Number of metrics (N): {N}")
print(f"Friedman chi² = {chi2_f:.4f}")

print("\n=== FF-statistic Approximation ===")
print(f"FF = {F_F:.4f}")
print(f"Degrees of Freedom: df1 = {df1}, df2 = {df2}")
print(f"Critical F-value (α = {alpha}): {F_critical:.4f}")
print(f"p-value = {p_value:.2e}")
print(f"Reject H₀: {reject_null}")
```
